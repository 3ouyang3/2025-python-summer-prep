## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
def count_characters(filename):
    char_count = {}
    
    try:
        with open(filename, 'r') as file:
            for line in file:
                for char in line.upper():
                    if char in char_count:
                        char_count[char] += 1
                    else:
                        char_count[char] = 1
                        
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    
    return char_count

file_path = 'lil-wayne.txt'  

result = count_characters(file_path)
if result:
    print(result)
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
from collections import Counter

def count_and_sort_characters(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read().upper()
            char_counts = Counter(content)
            
            sorted_counts = sorted(char_counts.items(), key=lambda item: item[1], reverse=True)
            
            return sorted_counts
            
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None

file_path = 'Week-03/intermediate/data/lil-wayne.txt'
result = count_and_sort_characters(file_path)

if result:
    print("Top 10 most frequent characters:")
    for char, count in result[:10]:
        print(f"'{char}': {count}")
    
    most_common_char, most_common_count = result[0]
    print(f"\nMost frequent character: '{most_common_char}' (appears {most_common_count} times)")
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

reverse_list2 = list2[::-1]

for x, y in zip(list1, reverse_list2):
    print(x, y)
````

## #10 Remove empty strings from the list of strings.
Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`

Expected Output:  `['I', 'love', 'data', 'science', '!']`

````python
list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']

def empty_string(strings):
    result = []
    for n in strings:
        if n != '': 
            result.append(n)
    return result

print(empty_string(list_of_strings))
````
